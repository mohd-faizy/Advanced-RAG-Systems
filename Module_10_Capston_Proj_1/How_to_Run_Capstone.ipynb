### Course alignment and free-first stack

- Covers: Capstone setup and execution guide for the production RAG components.
- Runtime stack: Groq chat models for generation/evaluation when an LLM is needed, plus local Hugging Face sentence-transformers embeddings for retrieval.
- No paid OpenAI API key is required. Set `GROQ_API_KEY` only for notebooks that call an LLM; pure retrieval and embedding notebooks run locally after model weights are available.
- Current LangChain pattern: provider split packages such as `langchain_groq`, `langchain_huggingface`, and `langchain_chroma`, with runnable `.invoke()` APIs.


## 🛠️ Detailed Component Breakdown

### 🎓 Main Notebook: `Production_RAG_Capstone.ipynb` 
The primary, interactive end-to-end project. It demonstrates the complete workflow from ingestion to evaluation within a single notebook environment, ideal for understanding how all modules fit together.

### 📥 Ingestion Module: `ingestion.py` 
Handles the data preparation layer. It is responsible for loading various document formats (PDF, TXT, CSV, Web), splitting them into semantic chunks, enriching metadata with tracking IDs, and building the initial Chroma vector index.

### 🔍 Retrieval Engine: `retrieval.py` 
Implements our high-performance **Hybrid Search**. It combines sparse keyword matching (BM25) with dense semantic search and applies a Cross-Encoder re-ranker to ensure the LLM receives the most precise context possible.

### 🧠 Agentic Orchestrator: `pipeline.py` 
The "brains" of the system built with **LangGraph**. It defines a state machine that classifies query complexity, grades retrieved documents for relevance, and autonomously decides when to rewrite a query for better results.

### 📊 Evaluation Suite: `evaluation.py` 
Our quality assurance framework. It implements an **LLM-as-Judge** pattern to automatically score the pipeline's performance on metrics like Faithfulness and Relevancy, ensuring the system remains grounded and helpful.

## Step 1: Clone the Repository

```bash
git clone https://github.com/mohd-faizy/Advanced-RAG-Systems.git
cd Advanced-RAG-Systems
```

## Step 2: Install Dependencies

Recommended with `uv`:

```bash
uv sync
```

Pip fallback:

```bash
python -m venv .venv
.venv\Scripts\activate
pip install -r requirements.txt
```

## Step 3: Set Up Your `.env` File

Copy the template into the project root:

```bash
copy .env.example .env
```

Minimum LLM configuration:

```env
GROQ_API_KEY=gsk_your_key_here
GROQ_MODEL=llama-3.1-8b-instant
EMBEDDING_MODEL=sentence-transformers/all-MiniLM-L6-v2
```

Embedding-only notebooks do not need a Groq key. They use local sentence-transformers models after the weights are downloaded.

## Step 4: Run the Capstone Notebook

```bash
uv run jupyter lab Module_10_Capstone/Production_RAG_Capstone.ipynb
```

Then run each cell top-to-bottom.

## Step 5: Run the Standalone Python Modules

```bash
# Build a local Chroma index from sample documents
uv run python Module_10_Capstone/ingestion.py --docs_dir ./data --collection my_rag

# Run the retrieval demo
uv run python Module_10_Capstone/retrieval.py

# Print the production best-practices checklist
uv run python Module_10_Capstone/evaluation.py
```

## Troubleshooting

| Issue | Solution |
| :--- | :--- |
| `GROQ_API_KEY` not found | Put `.env` in the project root, not inside `Module_10_Capstone/` |
| `ModuleNotFoundError: langchain_chroma` | Run `uv sync` or `pip install -r requirements.txt` |
| First embedding run is slow | The local embedding model is downloading; later runs use the cache |
| Groq rate limit errors | Wait briefly, reduce batch size, or switch `GROQ_MODEL` to another available Groq model |
| Chroma DB lock errors | Stop other notebook kernels using the DB, then rerun |
