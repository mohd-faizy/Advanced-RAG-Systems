## 🛠️ Detailed Component Breakdown

### 🎓 Main Notebook: `Production_RAG_Capstone.ipynb` 
The primary, interactive end-to-end project. It demonstrates the complete workflow from ingestion to evaluation within a single notebook environment, ideal for understanding how all modules fit together.

### 📥 Ingestion Module: `ingestion.py` 
Handles the data preparation layer. It is responsible for loading various document formats (PDF, TXT, CSV, Web), splitting them into semantic chunks, enriching metadata with tracking IDs, and building the initial Chroma vector index.

### 🔍 Retrieval Engine: `retrieval.py` 
Implements our high-performance **Hybrid Search**. It combines sparse keyword matching (BM25) with dense semantic search and applies a Cross-Encoder re-ranker to ensure the LLM receives the most precise context possible.

### 🧠 Agentic Orchestrator: `pipeline.py` 
The "brains" of the system built with **LangGraph**. It defines a state machine that classifies query complexity, grades retrieved documents for relevance, and autonomously decides when to rewrite a query for better results.

### 📊 Evaluation Suite: `evaluation.py` 
Our quality assurance framework. It implements an **LLM-as-Judge** pattern to automatically score the pipeline's performance on metrics like Faithfulness and Relevancy, ensuring the system remains grounded and helpful.

## Step 1: Clone the Repository

```bash
git clone https://github.com/mohd-faizy/Advanced-RAG-Systems.git
cd Advanced-RAG-Systems
```

## Step 2: Create a Virtual Environment

```bash
# Windows
python -m venv .venv
.venv\Scripts\activate

# macOS / Linux
python3 -m venv .venv
source .venv/bin/activate
```

## Step 3: Install Dependencies

```bash
pip install langchain langchain-groq langchain-huggingface langchain-community \
            langchain-text-splitters langgraph chromadb sentence-transformers \
            rank-bm25 python-dotenv jupyter ragas
```

## Step 4: Set Up Your `.env` File

Create a `.env` file in the **project root** (same folder as this notebook's parent):

```env
GROQ_API_KEY=gsk_your_key_here
```

> **How to get a Groq API Key (free):**
> 1. Go to [console.groq.com](https://console.groq.com)
> 2. Sign up / Log in
> 3. Navigate to **API Keys** → **Create API Key**
> 4. Copy the key and paste it into your `.env` file

## Step 5: Run the Capstone Notebook

### Option A: Jupyter Notebook (Interactive)
```bash
jupyter notebook Module_10_Capstone/Production_RAG_Capstone.ipynb
```
Then run each cell top-to-bottom using `Shift+Enter`.

### Option B: Command Line (Automated)
```bash
jupyter nbconvert --execute --to notebook --inplace \
    Module_10_Capstone/Production_RAG_Capstone.ipynb
```

## Step 6: Run the Standalone Python Modules

The capstone also includes 4 standalone Python scripts that can be used independently:

```bash
# Run the ingestion pipeline on a folder of documents
python Module_10_Capstone/ingestion.py --docs_dir ./data --collection my_rag

# Run the retrieval demo
python Module_10_Capstone/retrieval.py

# Print the production best-practices checklist
python Module_10_Capstone/evaluation.py
```

## Troubleshooting

| Issue | Solution |
| :--- | :--- |
| `GROQ_API_KEY` not found | Ensure `.env` is in the project root, not inside `Module_10_Capstone/` |
| `ModuleNotFoundError` | Run `pip install -r requirements.txt` or the install command above |
| Rate limit errors (429) | Wait 30 seconds and re-run. Groq free tier = 6000 TPM |
| `TOKENIZERS_PARALLELISM` warning | Already suppressed in code via `os.environ` |
| Chroma DB lock errors | Delete the `chroma_capstone_db/` folder and re-run |
